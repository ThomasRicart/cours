## Partie 1 – Identité numérique : qui es-tu en ligne ? *(20 min)*

### 📖 Définitions

**Identité numérique** : ensemble des traces, informations et contenus qu'une personne laisse ou publie sur Internet. Elle se construit à partir de ce que l'on choisit de montrer **mais aussi** de ce que les autres publient à notre sujet.

**E-réputation** : image que les autres ont de toi en ligne. Elle peut être construite par toi-même (tes posts, photos, commentaires) ou par d'autres (tags, mentions, articles...)

**Identification** : déclarer qui on est (ex : entrer son nom d'utilisateur). C'est dire *"je suis Léa Dupont"*.

**Authentification** : prouver qu'on est bien celui qu'on prétend être (ex : entrer un mot de passe, recevoir un SMS de validation). C'est *prouver* qu'on est Léa Dupont.

---
### 🔐 Identification vs Authentification : un exemple concret

| Étape | Action | Concept |
|-------|--------|--------|
| Je tape mon email | Je *déclare* qui je suis | **Identification** |
| Je tape mon mot de passe | Je *prouve* que c'est bien moi | **Authentification** |
| Je reçois un SMS de confirmation | 2ème preuve (double authentification) | **Authentification forte** |

## Partie 3 – Le modèle économique *(25 min)*

### 💰 "Quand c'est gratuit, c'est toi le produit"

Les réseaux sociaux sont **gratuits** pour les utilisateurs. Mais comment ces entreprises gagnent-elles des milliards d'euros ?

Voici un schéma simplifié du modèle économique :

```
  Tu t'inscris        Tu utilises           Tu génères
  gratuitement   →    le service     →      des données
                                              ↓
  Annonceurs    ←    Publicité       ←    Données analysées
  paient Meta        ciblée               (âge, goûts, localisation...)
```

### 📊 Quelques chiffres
- En 2023, Meta (Facebook + Instagram) a réalisé **~117 milliards $** de chiffre d'affaires
- **97%** de ces revenus proviennent de la publicité
- Un utilisateur Facebook génère en moyenne **~50€/an** de revenus publicitaires pour Meta

**Scénario :** Mathieu, 16 ans, utilise Instagram. Il regarde souvent des vidéos de basketball, like des posts de sneakers, et suit plusieurs joueurs NBA. Il habite à Lyon.

1. Quelles données Instagram collecte-t-il sur Mathieu ?

4. Est-il légal de cibler des publicités vers des mineurs ? Recherche la réglementation en vigueur.
5. Que dit le **RGPD** (Règlement Général sur la Protection des Données) sur la collecte de données des mineurs de moins de 15 ans ?

### ✏️ Exercice 4 – Lire les CGU

Les **Conditions Générales d'Utilisation** (CGU) sont le contrat que tu acceptes quand tu crées un compte sur un réseau social. Presque personne ne les lit... et pourtant !

Recherche et lis (au moins en partie) les CGU d'un réseau social de ton choix.

1. Quel est l'**âge minimum** pour s'inscrire ?
2. Qui est **propriétaire** des photos que tu publies ?
3. Le réseau peut-il **supprimer ton compte** sans te prévenir ? Dans quels cas ?
4. Pendant combien de temps tes données sont-elles conservées après suppression du compte ?
5. Note une clause qui t'a surpris(e).

## Partie 4 – Paramétrer sa confidentialité *(10 min)*

### 🔒 Activité pratique

Sur un réseau social de ton choix (ou en utilisant un compte fictif de démonstration), explore les paramètres de confidentialité.

Réponds aux questions suivantes :

1. Par défaut, qui peut voir tes publications ? (Tout le monde / Amis / Moi seul)
2. Comment limiter la visibilité de tes publications à tes amis seulement ?
3. Comment empêcher qu'on te tague en photo sans ta permission ?
4. Comment savoir quelles applications tierces ont accès à ton compte ?
5. Comment télécharger toutes les données que le réseau a collectées sur toi ?

*(Note : sur Facebook, cette option s'appelle "Télécharger mes informations". Sur Instagram : "Télécharger mes données".)*

## 📝 Résumé de la séance

| Notion | Définition |
|--------|------------|
| **Identité numérique** | Ensemble des traces qu'on laisse ou qu'on publie en ligne |
| **E-réputation** | Image que les autres ont de toi sur Internet |
| **Identification** | Déclarer qui on est (login, email) |
| **Authentification** | Prouver qu'on est bien qui on prétend être (mot de passe, 2FA) |
| **Modèle économique** | Les réseaux sociaux financent leur gratuité par la publicité ciblée |
| **CGU** | Contrat à accepter pour utiliser un service numérique |
