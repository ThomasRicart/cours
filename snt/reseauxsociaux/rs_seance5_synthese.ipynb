## Mission 2 – Analyse du graphe d'amitié *(20 min)*

```python
# 2a. Qui a le plus d'amis ? (degré maximal)
print("=== NOMBRE D'AMIS PAR UTILISATEUR ===")
for uid, liste_amis in sorted(amis.items(), key=lambda x: len(x[1]), reverse=True):
    pseudo = utilisateurs[uid]["pseudo"]
    print(f"  {pseudo:12} ({uid}) : {len(liste_amis)} ami(s)")
```

```python
# 2b. Calcul du diamètre du graphe
# (réutilisation des fonctions de la séance 2)
from collections import deque

def distance(graphe, depart, arrivee):
    if depart == arrivee:
        return 0
    file = deque([(depart, 0)])
    visites = {depart}
    while file:
        sommet, dist = file.popleft()
        for voisin in graphe[sommet]:
            if voisin == arrivee:
                return dist + 1
            if voisin not in visites:
                visites.add(voisin)
                file.append((voisin, dist + 1))
    return float('inf')

def excentricite(graphe, sommet):
    distances = [distance(graphe, sommet, autre) for autre in graphe]
    distances_finies = [d for d in distances if d != float('inf')]
    return max(distances_finies) if distances_finies else float('inf')

# Calcul
excs = {uid: excentricite(amis, uid) for uid in amis}
rayon = min(v for v in excs.values() if v != float('inf'))
diametre = max(v for v in excs.values() if v != float('inf'))
centre = [uid for uid, exc in excs.items() if exc == rayon]

print(f"Rayon    : {rayon}")
print(f"Diamètre : {diametre}")
print(f"Centre   : {[utilisateurs[uid]['pseudo'] for uid in centre]}")
print()
print("Interprétation : dans ce réseau, deux personnes quelconques sont au maximum à", diametre, "connexions l'une de l'autre.")
```

```python
# 2d. Recommandations d'amis pour U01

def amis_communs(graphe, p1, p2):
    return [a for a in graphe[p1] if a in graphe[p2]]

def recommander(graphe, uid, n=3):
    scores = {}
    for candidat in graphe:
        if candidat == uid or candidat in graphe[uid]:
            continue
        nb = len(amis_communs(graphe, uid, candidat))
        if nb > 0:
            scores[candidat] = nb
    return sorted(scores.items(), key=lambda x: x[1], reverse=True)[:n]

print(f"=== RECOMMANDATIONS POUR {utilisateurs['U01']['pseudo']} ===")
for uid_rec, score in recommander(amis, "U01"):
    print(f"  → {utilisateurs[uid_rec]['pseudo']} ({score} ami(s) commun(s))")
```
