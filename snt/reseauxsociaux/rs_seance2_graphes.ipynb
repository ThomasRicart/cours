## Partie 1 – Un graphe, c'est quoi ? *(15 min)*

### 📖 Vocabulaire de base

Un **graphe** est une façon de représenter des relations entre des objets.

- Les **sommets** (ou nœuds) représentent les **personnes** (ou entités)
- Les **arêtes** (ou liens) représentent les **relations** entre elles (amitié, abonnement...)

**Exemple : le graphe d'amitié d'une classe**

```
    Alice ──── Bob ──── Chloé
      │          │
    David ──── Emma ──── Farid
                │
              Gaël
```

Dans ce graphe :
- Alice et Bob sont **amis** (il y a une arête entre eux)
- Alice et Chloé **ne sont pas directement amis** (pas d'arête directe)
- Le **degré** d'un sommet = nombre de liens qu'il possède (Bob a un degré de 3)

## Partie 2 – Rayon, diamètre et centre d'un graphe *(30 min)*

### 📐 Définitions

Pour mesurer les propriétés d'un graphe, on utilise des notions de **distance** :

**Distance entre deux sommets** = nombre minimum d'arêtes à parcourir pour aller de l'un à l'autre.

**Exemple :** Dans le graphe ci-dessus, la distance de Alice à Chloé est **2** (Alice → Bob → Chloé).

---

**Excentricité d'un sommet** = distance maximale entre ce sommet et tous les autres.

**Rayon du graphe** = plus petite excentricité (le sommet le mieux placé)

**Diamètre du graphe** = plus grande excentricité (la plus grande distance entre deux sommets quelconques)

**Centre du graphe** = sommet(s) dont l'excentricité est égale au rayon

## Partie 3 – La notion de "petit monde" *(20 min)*

### 🧪 L'expérience de Milgram (1967)

En 1967, le psychologue américain Stanley Milgram a réalisé une expérience fascinante :

Il a envoyé des lettres à des personnes **aléatoires** dans le Nebraska (États-Unis) avec pour mission de faire parvenir la lettre à une personne **précise** à Boston, en la transmettant uniquement de proche en proche (à des gens qu'ils connaissaient personnellement).

**Résultat :** En moyenne, il fallait seulement **6 intermédiaires** pour atteindre n'importe quelle personne.

👉 C'est l'origine de l'expression **"six degrés de séparation"** : tout être humain sur Terre serait connecté à n'importe quel autre par une chaîne de 6 relations maximum.

## Partie 4 – Explorer un graphe avec Python *(20 min)*

On peut utiliser Python pour représenter et analyser un graphe !

````python
# Représentation d'un graphe sous forme de dictionnaire d'adjacence
# Pour chaque personne, on liste ses amis

graphe = {
    "Alice":  ["Bob", "David"],
    "Bob":    ["Alice", "Chloé", "Emma"],
    "Chloé":  ["Bob", "Farid"],
    "David":  ["Alice", "Emma"],
    "Emma":   ["Bob", "David", "Gaël"],
    "Farid":  ["Chloé"],
    "Gaël":   ["Emma"],
}

# Affichage du graphe
print("=== GRAPHE D'AMITIÉ ===")
for personne, amis in graphe.items():
    print(f"{personne:8} ── amis avec : {', '.join(amis)}")
````

````python
# Calcul du degré de chaque sommet
print("=== DEGRÉ DE CHAQUE SOMMET ===")
for personne, amis in graphe.items():
    print(f"{personne:8} : degré {len(amis)}")
````

````python
# Algorithme de calcul de distance entre deux sommets (BFS - parcours en largeur)
from collections import deque

def distance(graphe, depart, arrivee):
    """Calcule la distance (chemin le plus court) entre deux sommets"""
    if depart == arrivee:
        return 0
    
    # On utilise une file (deque) pour explorer le graphe niveau par niveau
    file = deque([(depart, 0)])  # (sommet actuel, distance depuis départ)
    visites = {depart}
    
    while file:
        sommet, dist = file.popleft()
        for voisin in graphe[sommet]:
            if voisin == arrivee:
                return dist + 1
            if voisin not in visites:
                visites.add(voisin)
                file.append((voisin, dist + 1))
    
    return float('inf')  # pas de chemin trouvé

# Tests
print("Distance Alice → Chloé :", distance(graphe, "Alice", "Chloé"))
print("Distance Alice → Gaël  :", distance(graphe, "Alice", "Gaël"))
print("Distance Farid → David :", distance(graphe, "Farid", "David"))
````

````python
# Calcul de la matrice complète des distances
personnes = list(graphe.keys())

print("=== MATRICE DES DISTANCES ===")
print(f"{'':8}", end="")
for p in personnes:
    print(f"{p[:5]:6}", end="")
print()
print("-" * (8 + 6 * len(personnes)))

for p1 in personnes:
    print(f"{p1:8}", end="")
    for p2 in personnes:
        d = distance(graphe, p1, p2)
        print(f"{d:6}", end="")
    print()
````

````python
# Calcul de l'excentricité, du rayon, du diamètre et du centre

def excentricite(graphe, sommet):
    """Retourne l'excentricité d'un sommet (distance max vers tous les autres)"""
    return max(distance(graphe, sommet, autre) for autre in graphe)

# Calcul pour chaque sommet
excentricites = {}
print("=== EXCENTRICITÉS ===")
for p in personnes:
    exc = excentricite(graphe, p)
    excentricites[p] = exc
    print(f"{p:8} : excentricité = {exc}")

rayon = min(excentricites.values())
diametre = max(excentricites.values())
centre = [p for p, exc in excentricites.items() if exc == rayon]

print()
print(f"Rayon    : {rayon}")
print(f"Diamètre : {diametre}")
print(f"Centre   : {centre}")
````

## 📝 Résumé

| Notion | Définition |
|--------|------------|
| **Graphe** | Ensemble de sommets reliés par des arêtes |
| **Degré** | Nombre de liens d'un sommet |
| **Distance** | Nombre minimum d'arêtes pour relier deux sommets |
| **Excentricité** | Distance maximale d'un sommet vers tous les autres |
| **Diamètre** | Plus grande excentricité du graphe |
| **Rayon** | Plus petite excentricité du graphe |
| **Centre** | Sommet(s) dont l'excentricité = rayon |
| **Petit monde** | Tout individu est à ~6 liens d'un autre (Milgram, 1967) |
