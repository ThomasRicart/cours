## Partie 1 – Comment les réseaux sociaux décident ce que tu vois *(20 min)*

### 📖 Les algorithmes de recommandation

Sur un réseau social, des millions de contenus sont publiés chaque seconde. Il est impossible de tout voir. C'est l'**algorithme de recommandation** qui décide ce que tu vois dans ton fil d'actualité.

Cet algorithme analyse notamment :
- Tes **interactions passées** : ce que tu as liké, partagé, commenté
- Le **temps passé** à regarder certains contenus
- Les **personnes que tu suis** et leurs interactions
- Ta **localisation** géographique
- L'**heure** de connexion
- Les **publicités** que tu as cliquées

**Objectif déclaré :** te montrer ce qui t'intéresse pour que tu restes le plus longtemps possible sur la plateforme.

**Objectif économique :** plus tu restes longtemps, plus tu vois de publicités, plus la plateforme gagne de l'argent.

## Partie 2 – La bulle de filtre *(20 min)*

### 🫧 Qu'est-ce qu'une bulle de filtre ?

La **bulle de filtre** (filter bubble) est un phénomène décrit par Eli Pariser en 2011 : en cherchant à nous montrer ce qui nous plaît, les algorithmes nous enferment progressivement dans une "bulle" où nous ne voyons plus que des contenus qui confirment nos opinions et nos goûts.

**Le mécanisme :**
```
Tu aimes X  →  L'algo te montre plus de X
           →  Tu interagis encore plus avec X
           →  L'algo te montre encore plus de X
           →  Tu ne vois plus que X
           →  Tu penses que tout le monde aime X
```

### 🤝 Bonding vs Bridging

| | **Bonding** | **Bridging** |
|---|---|---|
| **Définition** | Renforcement des liens existants au sein d'un groupe | Construction de nouveaux liens avec des personnes différentes |
| **Effet** | Sentiment d'appartenance, cohésion de groupe | Ouverture sur la diversité, échange d'idées nouvelles |
| **Risque** | Enfermement, pensée de groupe | / |
| **Exemple réseau social** | Tu suis uniquement des gens qui partagent tes goûts | Tu découvres des personnes aux idées différentes |

Les algorithmes de recommandation favorisent massivement le **bonding** au détriment du **bridging**.

## Partie 3 – Coder un algorithme de recommandation simple *(35 min)*

On va construire un algorithme de recommandation d'amis basé sur les **amis communs** : "Tu pourrais connaître X car vous avez 3 amis en commun."

````python
# Notre réseau social : qui est ami avec qui ?
reseau = {
    "Alice":   ["Bob", "David", "Emma"],
    "Bob":     ["Alice", "Chloé", "Emma", "Hugo"],
    "Chloé":   ["Bob", "Farid", "Inès"],
    "David":   ["Alice", "Emma", "Julien"],
    "Emma":    ["Alice", "Bob", "David", "Gaël"],
    "Farid":   ["Chloé", "Inès"],
    "Gaël":    ["Emma", "Hugo"],
    "Hugo":    ["Bob", "Gaël", "Inès"],
    "Inès":    ["Chloé", "Farid", "Hugo"],
    "Julien":  ["David"],
}

print("Réseau social chargé !")
print(f"Nombre d'utilisateurs : {len(reseau)}")
````

````python
# Algorithme : compter les amis communs entre deux personnes

def amis_communs(reseau, personne1, personne2):
    """Retourne la liste des amis communs entre deux personnes"""
    amis1 = reseau[personne1]  # liste des amis de personne1
    amis2 = reseau[personne2]  # liste des amis de personne2
    
    communs = []
    for ami in amis1:
        if ami in amis2:       # si cet ami est aussi ami de personne2
            communs.append(ami)
    
    return communs

# Test
communs = amis_communs(reseau, "Alice", "Chloé")
print(f"Amis communs entre Alice et Chloé : {communs} ({len(communs)} en commun)")

communs2 = amis_communs(reseau, "Alice", "Hugo")
print(f"Amis communs entre Alice et Hugo  : {communs2} ({len(communs2)} en commun)")
````

````python
# Algorithme de recommandation : suggérer des amis potentiels

def recommander_amis(reseau, personne, nb_recommandations=3):
    """Recommande des personnes à ajouter en ami"""
    amis_actuels = reseau[personne]
    scores = {}  # on va stocker le score de chaque candidat
    
    for candidat in reseau:
        # On ne recommande pas la personne elle-même
        if candidat == personne:
            continue
        # On ne recommande pas quelqu'un déjà ami
        if candidat in amis_actuels:
            continue
        
        # Le score = nombre d'amis communs
        nb_communs = len(amis_communs(reseau, personne, candidat))
        if nb_communs > 0:
            scores[candidat] = nb_communs
    
    # On trie par score décroissant
    recommandations = sorted(scores.items(), key=lambda x: x[1], reverse=True)
    
    return recommandations[:nb_recommandations]

# Test pour Alice
print("=== RECOMMANDATIONS POUR ALICE ===")
for personne, score in recommander_amis(reseau, "Alice"):
    communs = amis_communs(reseau, "Alice", personne)
    print(f"  👤 {personne} ({score} ami(s) commun(s) : {', '.join(communs)})")
````

````python
# Affichons les recommandations pour tout le monde
print("=== RECOMMANDATIONS POUR TOUS LES UTILISATEURS ===")
for utilisateur in reseau:
    recs = recommander_amis(reseau, utilisateur)
    if recs:
        noms = [f"{p} ({s} commun)" for p, s in recs]
        print(f"{utilisateur:8} → {', '.join(noms)}")
    else:
        print(f"{utilisateur:8} → aucune recommandation")
````

## Partie 4 – Vrai ou faux ? Détecter la désinformation *(10 min)*

### 🔍 Les fake news et les réseaux sociaux

Les algorithmes de recommandation favorisent les contenus **engageants**, c'est-à-dire ceux qui génèrent des réactions (likes, partages, commentaires). Or, les fausses informations sont souvent plus **choquantes ou émotionnelles** que les vraies, donc elles génèrent plus d'engagement.

**Résultat :** Les algorithmes propagent involontairement (mais structurellement) la désinformation.
