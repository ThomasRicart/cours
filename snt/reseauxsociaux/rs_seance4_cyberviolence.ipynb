## Partie 1 – Les formes de cyberviolence *(25 min)*

### 📖 Définitions

La **cyberviolence** désigne toute forme de violence exercée via les outils numériques (réseaux sociaux, messageries, jeux en ligne, e-mails...).

| Forme | Définition | Exemple |
|-------|------------|--------|
| **Cyberharcèlement** | Envoi répété de messages hostiles, insultes, menaces | Insultes répétées sur Instagram, "raid" organisé contre quelqu'un |
| **Discrimination** | Messages haineux visant une caractéristique (race, religion, orientation...) | Commentaires racistes sous une publication |
| **Sexting non consenti** | Partage de photos ou vidéos intimes sans accord de la personne | Partager des photos d'un(e) ex sans son consentement |
| **Usurpation d'identité** | Créer un faux profil au nom d'une autre personne | Créer un compte avec le nom et la photo d'un camarade |
| **Doxing** | Publier des informations personnelles (adresse, téléphone...) pour nuire | Publier l'adresse d'une personne pour que d'autres la harcèlent |
| **Flaming** | Échanges violents et agressifs en ligne | "Flame war" dans les commentaires |
| **Exclusion** | Exclure délibérément quelqu'un d'un groupe en ligne | Retirer quelqu'un d'un groupe WhatsApp de classe |
| **Grooming** | Manipulation d'un mineur par un adulte en vue d'une relation abusive | Un adulte qui prétend être un ado pour approcher des mineurs |

## Partie 2 – Ce que dit la loi *(20 min)*

### ⚖️ Le cadre juridique

La cyberviolence n'est pas seulement un problème moral : c'est aussi une affaire **légale**.

---

#### Article 222-33-2-2 du Code pénal – Le cyberharcèlement

> *"Le fait de harceler une personne par des propos ou comportements répétés ayant pour objet ou pour effet une dégradation de ses conditions de vie se traduisant par une altération de sa santé physique ou mentale est puni de deux ans d'emprisonnement et de 30 000 € d'amende lorsque ces faits ont causé une incapacité totale de travail inférieure ou égale à huit jours ou n'ont entraîné aucune incapacité de travail."*

**Points clés à retenir :**
- Le cyberharcèlement est **puni par la loi**
- Il faut un caractère **répété** (plusieurs actes, pas juste un)
- La peine peut aller jusqu'à **3 ans et 45 000 €** si la victime est un mineur
- Même **être complice** (liker, partager, ne pas intervenir tout en faisant partie du groupe) peut engager la responsabilité
- L'**âge des auteurs** ne les protège pas totalement : les mineurs peuvent être jugés et sanctionnés

---

#### Autres textes de loi pertinents

| Infraction | Loi | Peine maximale |
|-----------|-----|---------------|
| Usurpation d'identité numérique | Article 226-4-1 Code pénal | 1 an, 15 000 € |
| Partage d'images intimes sans consentement | Article 226-2-1 Code pénal | 2 ans, 60 000 € |
| Incitation à la haine | Article 24 Loi de 1881 | 1 an, 45 000 € |
| Grooming (sollicitation de mineur) | Article 227-22-1 Code pénal | 2 ans, 30 000 € |

## Partie 3 – Se protéger et agir *(25 min)*

### 🛡️ Outils techniques de protection

**Sur les réseaux sociaux, tu peux :**
- **Bloquer** une personne (elle ne peut plus te contacter ni voir ton profil)
- **Signaler** un contenu ou un compte (les modérateurs examinent le signalement)
- **Rendre ton compte privé** (seuls tes abonnés approuvés voient ton contenu)
- **Filtrer les commentaires** avec des mots-clés bloqués
- **Désactiver** les demandes de messages de personnes inconnues

## Partie 4 – Activité : l'authenticité en ligne *(15 min)*

### 🤔 Peut-on faire confiance aux profils en ligne ?

Les réseaux sociaux posent de vraies questions sur l'**authenticité** :
- Un profil vérifié (✓ bleu) garantit-il que c'est bien la vraie personne ?
- Une photo peut-elle être retouchée au point de tromper sur l'identité ?
- Un compte avec des milliers d'abonnés est-il forcément influent ? Ou peut-il avoir acheté ses abonnés ?

### 🔍 Activité de vérification

**Outils de vérification d'image :**
- **Google Images** (recherche inversée par image) : [images.google.com](https://images.google.com)
- **TinEye** : [tineye.com](https://tineye.com)

**Questions pratiques :**

1. Fais une recherche inversée sur la photo de profil d'un compte que tu trouves suspect (ou utilise une image fournie par ton professeur). Qu'obtiens-tu ?

2. Un compte Instagram "lifestyle" a 50 000 abonnés mais ses photos ne font jamais plus de 20 likes. Qu'est-ce que cela peut indiquer ?

3. Qu'est-ce qu'un **deepfake** ? Recherche une définition et un exemple. Pourquoi est-ce dangereux sur les réseaux sociaux ?

4. Les plateformes proposent des "badges de vérification" (Instagram, X...). Qui peut en obtenir un ? Cela garantit-il que le contenu publié est vrai ?

## 📝 Résumé et ressources

### Les formes de cyberviolence
Cyberharcèlement, discrimination, sexting non consenti, usurpation d'identité, doxing, grooming...

### La loi
La cyberviolence est punie par le Code pénal. L'article 222-33-2-2 punit le cyberharcèlement jusqu'à 3 ans de prison (aggravé si victime mineure). L'âge des auteurs ne les immunise pas.

### En cas de cyberviolence
1. **Conserver des preuves** (captures d'écran avec date)
2. **Bloquer et signaler** sur la plateforme
3. **En parler à un adulte** de confiance
4. **Appeler le 3018** ou visiter nonauharcelement.education.gouv.fr
5. **Porter plainte** si nécessaire
