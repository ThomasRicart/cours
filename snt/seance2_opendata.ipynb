## Partie 1 – C'est quoi l'Open Data ? *(15 min)*

### 📖 Des données ouvertes au public

L'**Open Data** (ou données ouvertes) désigne des données que leurs producteurs mettent librement à disposition de tout le monde, gratuitement, pour que chacun puisse les consulter, les réutiliser et les redistribuer.

**Qui publie des données ouvertes ?**
- L'État et les collectivités locales (mairies, régions...)
- L'Union Européenne
- Des organisations internationales (OMS, ONU...)
- Certaines entreprises et associations

**Quelques exemples de portails Open Data :**
- 🇫🇷 [data.gouv.fr](https://www.data.gouv.fr) – portail officiel des données publiques françaises
- 🇪🇺 [data.europa.eu](https://data.europa.eu) – données de l'Union Européenne
- 🌍 [ourworldindata.org](https://ourworldindata.org) – données mondiales (santé, environnement...)

## Partie 2 – Chargement d'un jeu de données réel *(20 min)*

### 📦 Le jeu de données de la séance

Pour cette séance, nous allons travailler avec un jeu de données sur les **prénoms les plus donnés en France** (source : INSEE, data.gouv.fr).

On va créer un extrait représentatif directement dans Python.

```python
# Création d'un fichier CSV de prénoms (extrait inspiré des données INSEE)
import csv

donnees_prenoms = [
    ["prenom", "sexe", "annee", "nombre"],
    ["Emma", "F", 2020, 5012],
    ["Jade", "F", 2020, 4821],
    ["Louise", "F", 2020, 4650],
    ["Alice", "F", 2020, 4200],
    ["Lina", "F", 2020, 3980],
    ["Gabriel", "M", 2020, 5230],
    ["Léo", "M", 2020, 4950],
    ["Raphaël", "M", 2020, 4780],
    ["Arthur", "M", 2020, 4560],
    ["Louis", "M", 2020, 4320],
    ["Emma", "F", 2010, 6120],
    ["Jade", "F", 2010, 5240],
    ["Inès", "F", 2010, 4870],
    ["Chloé", "F", 2010, 4650],
    ["Léa", "F", 2010, 4400],
    ["Enzo", "M", 2010, 5890],
    ["Lucas", "M", 2010, 5640],
    ["Mathis", "M", 2010, 5310],
    ["Hugo", "M", 2010, 5080],
    ["Tom", "M", 2010, 4750],
    ["Emma", "F", 2000, 7850],
    ["Léa", "F", 2000, 7200],
    ["Manon", "F", 2000, 6900],
    ["Camille", "F", 2000, 6400],
    ["Julie", "F", 2000, 6100],
    ["Théo", "M", 2000, 7100],
    ["Thomas", "M", 2000, 6800],
    ["Nicolas", "M", 2000, 6500],
    ["Maxime", "M", 2000, 6200],
    ["Antoine", "M", 2000, 5900],
]

with open("prenoms.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(donnees_prenoms)

print("Fichier prenoms.csv créé !")
```

```python
# Chargement et première exploration
import csv

prenoms = []
with open("prenoms.csv", encoding="utf-8") as f:
    lecteur = csv.DictReader(f)
    for ligne in lecteur:
        prenoms.append(ligne)

print(f"Nombre d'entrées : {len(prenoms)}")
print("\nPremière entrée :")
print(prenoms[0])
print("\nDescripteurs disponibles :")
print(list(prenoms[0].keys()))
```

```python
# ⚠️ Attention aux types de données !
# Quand on lit un CSV, TOUT est lu comme du texte (string)
# Il faut convertir si on veut faire des calculs

premier = prenoms[0]
print("Type du prénom :", type(premier["prenom"]))    # str (texte)
print("Type de l'année :", type(premier["annee"]))    # str aussi !
print("Type du nombre :", type(premier["nombre"]))    # str aussi !

print()
# Pour faire des calculs, on doit convertir :
annee_en_nombre = int(premier["annee"])  # conversion en entier
nombre_en_entier = int(premier["nombre"])
print(f"Après conversion : {premier['prenom']} est apparu {nombre_en_entier} fois en {annee_en_nombre}")
```

## Partie 3 – Premières explorations *(35 min)*

### 🔎 Recherche dans les données

```python
# Rechercher un prénom spécifique
prenom_cherche = "Emma"

print(f"Résultats pour '{prenom_cherche}' :")
for p in prenoms:
    if p["prenom"] == prenom_cherche:
        print(f"  {p['annee']} → {p['nombre']} naissances")
```

```python
# Tous les prénoms d'une année donnée
annee_choisie = "2020"

print(f"Prénoms populaires en {annee_choisie} :")
for p in prenoms:
    if p["annee"] == annee_choisie:
        print(f"  {p['prenom']} ({p['sexe']}) : {p['nombre']} naissances")
```

```python
# Calcul du total de naissances pour une année
annee_choisie = "2020"
total = 0

for p in prenoms:
    if p["annee"] == annee_choisie:
        total = total + int(p["nombre"])  # on additionne les naissances

print(f"Total de naissances (top 10) en {annee_choisie} : {total}")
```
