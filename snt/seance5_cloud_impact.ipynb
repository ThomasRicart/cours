### 📦 Les supports de stockage

| Support | Type | Capacité typique | Caractéristique |
|---------|------|------------------|-----------------|
| Disque dur (HDD) | Interne | 500 Go – 4 To | Mécanique, sensible aux chocs |

### ☁️ Le cloud, c'est quoi concrètement ?

Tu utilises sûrement déjà le cloud :
- 📸 Google Photos, iCloud (photos)
- 🎵 Spotify, Deezer (musique en streaming)
- 📁 Google Drive, OneDrive (documents)
- 💬 WhatsApp, Snapchat (messages)

## Partie 2 – Les centres de données *(15 min)*

### 🏭 Un datacenter, c'est quoi ?

Un **centre de données** est un bâtiment qui abrite des milliers de serveurs (ordinateurs puissants) qui stockent et traitent des données en permanence.

Pour fonctionner, un datacenter nécessite :
- ⚡ Une **alimentation électrique** continue et redondante (il ne peut jamais s'éteindre)
- ❄️ Un système de **refroidissement** puissant (les serveurs chauffent énormément)
- 💧 De l'**eau** (pour les systèmes de refroidissement par évaporation)
- 🔒 Une sécurité physique importante

### 📊 Quelques chiffres

- Les datacenters consomment environ **1 à 2 %** de l'électricité mondiale
- Un grand datacenter peut utiliser des **millions de litres d'eau** par an pour le refroidissement
- Le numérique représente environ **4 %** des émissions mondiales de gaz à effet de serre (et cette part augmente)
- Un email avec une pièce jointe de 1 Mo émet environ **19 grammes de CO₂**

## Partie 3 – Calculer l'impact de nos usages *(30 min)*

### 🔬 Activité Python : simulateur d'empreinte numérique

On va créer un programme pour estimer l'impact de différents usages numériques.

```python
# Données d'empreinte carbone de différents usages numériques
# (ordres de grandeur approximatifs, sources : ADEME, The Shift Project)

usages = {
    "Email simple (sans pièce jointe)": {"co2_g": 4, "unite": "email"},
    "Email avec pièce jointe (1 Mo)": {"co2_g": 19, "unite": "email"},
    "Recherche Google": {"co2_g": 0.2, "unite": "recherche"},
    "Streaming vidéo HD (1h)": {"co2_g": 36, "unite": "heure"},
    "Streaming musique (1h)": {"co2_g": 7, "unite": "heure"},
    "Visioconférence (1h)": {"co2_g": 157, "unite": "heure"},
    "Stocker 1 Go dans le cloud (1 an)": {"co2_g": 32, "unite": "Go"},
    "Post sur les réseaux sociaux avec photo": {"co2_g": 0.2, "unite": "post"},
    "Jeu vidéo en ligne (1h)": {"co2_g": 100, "unite": "heure"},
}

print("=== EMPREINTE CARBONE DES USAGES NUMÉRIQUES ===")
print(f"{'Usage':<45} {'CO₂ (g)':<10}")
print("-" * 55)
for usage, data in usages.items():
    print(f"{usage:<45} {data['co2_g']:<10} g CO₂ / {data['unite']}")
```

```python
# Calcul de l'empreinte d'une journée type
# Change les valeurs pour simuler ta propre journée !

ma_journee = {
    "Email simple (sans pièce jointe)": 10,       # 10 emails simples
    "Email avec pièce jointe (1 Mo)": 2,           # 2 emails avec pièce jointe
    "Recherche Google": 15,                         # 15 recherches
    "Streaming vidéo HD (1h)": 2,                  # 2h de Netflix
    "Streaming musique (1h)": 1,                   # 1h de musique
    "Visioconférence (1h)": 0,                     # pas de visio
    "Stocker 1 Go dans le cloud (1 an)": 0,        # déjà compté annuellement
    "Post sur les réseaux sociaux avec photo": 5,  # 5 posts
    "Jeu vidéo en ligne (1h)": 1,                  # 1h de jeu
}

total_co2 = 0
print("=== MA JOURNÉE NUMÉRIQUE ===")
for usage, quantite in ma_journee.items():
    if quantite > 0:
        co2_unitaire = usages[usage]["co2_g"]
        co2_total_usage = co2_unitaire * quantite
        total_co2 += co2_total_usage
        print(f"{usage[:40]:<40} × {quantite:>3} = {co2_total_usage:>8.1f} g CO₂")

print("-" * 60)
print(f"{'TOTAL journée':<45} {total_co2:>8.1f} g CO₂")
print(f"\nSoit l'équivalent de {total_co2 / 1000:.2f} kg de CO₂")
print(f"Ou encore {total_co2 / 1000 * 365:.0f} kg / an")
```

```python
# Visualisation de la répartition de l'empreinte
import matplotlib.pyplot as plt

labels = []
valeurs = []

for usage, quantite in ma_journee.items():
    if quantite > 0:
        co2 = usages[usage]["co2_g"] * quantite
        if co2 > 0:
            labels.append(usage[:30] + "..." if len(usage) > 30 else usage)
            valeurs.append(co2)

plt.figure(figsize=(9, 6))
plt.barh(labels, valeurs, color="steelblue")
plt.xlabel("CO₂ émis (grammes)")
plt.title("Mon empreinte numérique journalière")
plt.tight_layout()
plt.savefig("empreinte_numerique.png")
plt.show()
print("Graphique sauvegardé !")
```
