## Partie 1 – Qu'est-ce qu'une donnée ? *(15 min)*

### 📖 Un peu de vocabulaire

Imaginons que tu veuilles décrire les élèves de ta classe dans un tableau.

| Nom | Prénom | Age | Ville |
|-----|--------|-----|-------|
| Dupont | Léa | 15 | Lyon |
| Martin | Tom | 16 | Paris |
| Bernard | Jade | 15 | Lyon |

**Définitions :**
- Une **donnée** est une valeur qui décrit quelque chose : `Léa`, `15`, `Lyon` sont des données.
- Un **descripteur** (ou attribut) est le nom de la colonne : `Nom`, `Prénom`, `Age`, `Ville`.
- Une **collection** est l'ensemble des lignes qui partagent les mêmes descripteurs : ici, la liste des élèves.
- Quand les données sont organisées en tableau, on parle de **données structurées**.

## Partie 2 – Le format CSV *(20 min)*

### 📄 C'est quoi CSV ?

**CSV** signifie *Comma Separated Values* (valeurs séparées par des virgules).  
C'est un format de fichier texte très simple pour stocker un tableau de données.

Le tableau de films de l'exercice précédent s'écrirait ainsi en CSV :

```
Titre,Réalisateur,Année,Durée,Genre
Intouchables,Nakache & Toledano,2011,112,Comédie
Inception,Christopher Nolan,2010,148,SF
Parasite,Bong Joon-ho,2019,132,Thriller
```

👉 La **première ligne** contient les descripteurs (l'en-tête).  
👉 Chaque ligne suivante est un objet de la collection.  
👉 Les valeurs sont séparées par des **virgules** (ou parfois des **point-virgules**).

## Partie 3 – Lire un fichier CSV avec Python *(35 min)*

### 🐍 Création du fichier CSV

On va d'abord créer un petit fichier CSV directement depuis Python, puis le lire.

````python
# Étape 1 : On crée un fichier CSV de films
contenu_csv = """Titre,Realisateur,Annee,Duree,Genre
Intouchables,Nakache et Toledano,2011,112,Comedie
Inception,Christopher Nolan,2010,148,SF
Parasite,Bong Joon-ho,2019,132,Thriller
Le Roi Lion,Roger Allers,1994,88,Animation
Titanic,James Cameron,1997,194,Drame
Retour vers le futur,Robert Zemeckis,1985,116,SF
Amelie Poulain,Jean-Pierre Jeunet,2001,122,Comedie
"""

with open("films.csv", "w", encoding="utf-8") as f:
    f.write(contenu_csv)

print("Fichier films.csv créé avec succès !")
````

````python
# Étape 2 : On lit le fichier CSV avec le module csv de Python
import csv

films = []  # liste vide qui va accueillir nos films

with open("films.csv", encoding="utf-8") as f:
    lecteur = csv.DictReader(f)  # lit le CSV comme un dictionnaire
    for ligne in lecteur:
        films.append(ligne)  # on ajoute chaque film à la liste

# Affichons le premier film pour voir la structure
print("Premier film :")
print(films[0])
print()
print(f"Nombre total de films : {len(films)}")
````

````python
# Étape 3 : Afficher tous les titres de films
print("Liste des films :")
for film in films:
    print("-", film["Titre"])  # on accède à la valeur du descripteur "Titre"
````

````python
# Étape 4 : Afficher titre ET année
print("Films avec leur année de sortie :")
for film in films:
    print(f"{film['Titre']} ({film['Annee']})")
````

## Partie 4 – Bilan et métadonnées *(10 min)*

### Les métadonnées, c'est quoi ?

Les **métadonnées** sont des données qui décrivent d'autres données (le fichier lui-même).  
Par exemple, pour une photo :
- La **date** de prise de vue
- Les **coordonnées GPS** du lieu
- Le **modèle d'appareil** utilisé

Pour un fichier CSV, les métadonnées pourraient être : la date de création, l'auteur, la source des données...
