## Mise en place – Le jeu de données *(10 min)*

On va travailler avec une table de **stations météo** et leurs relevés.  
Ce type de données est disponible sur Météo-France en Open Data.

```python
# Création du jeu de données météo
import csv

meteo = [
    ["ville", "region", "mois", "temp_moy", "precip_mm", "ensoleillement_h"],
    ["Paris", "Ile-de-France", "Janvier", 4.5, 51, 62],
    ["Paris", "Ile-de-France", "Juillet", 20.1, 59, 217],
    ["Marseille", "PACA", "Janvier", 8.2, 43, 152],
    ["Marseille", "PACA", "Juillet", 25.3, 11, 340],
    ["Lyon", "Auvergne-Rhone-Alpes", "Janvier", 3.1, 47, 73],
    ["Lyon", "Auvergne-Rhone-Alpes", "Juillet", 22.8, 58, 248],
    ["Bordeaux", "Nouvelle-Aquitaine", "Janvier", 6.3, 89, 90],
    ["Bordeaux", "Nouvelle-Aquitaine", "Juillet", 21.4, 47, 252],
    ["Lille", "Hauts-de-France", "Janvier", 3.2, 68, 45],
    ["Lille", "Hauts-de-France", "Juillet", 18.5, 63, 195],
    ["Nice", "PACA", "Janvier", 9.5, 62, 160],
    ["Nice", "PACA", "Juillet", 26.1, 8, 348],
    ["Strasbourg", "Grand Est", "Janvier", 1.8, 38, 54],
    ["Strasbourg", "Grand Est", "Juillet", 20.9, 71, 225],
    ["Rennes", "Bretagne", "Janvier", 5.8, 92, 68],
    ["Rennes", "Bretagne", "Juillet", 18.2, 55, 208],
    ["Toulouse", "Occitanie", "Janvier", 5.9, 52, 108],
    ["Toulouse", "Occitanie", "Juillet", 23.5, 30, 288],
    ["Brest", "Bretagne", "Janvier", 7.1, 135, 58],
    ["Brest", "Bretagne", "Juillet", 17.3, 72, 190],
]

with open("meteo.csv", "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(meteo)

print("Fichier meteo.csv créé !")

# Chargement
releves = []
with open("meteo.csv", encoding="utf-8") as f:
    lecteur = csv.DictReader(f)
    for ligne in lecteur:
        releves.append(ligne)

print(f"{len(releves)} relevés chargés.")
print("Exemple :", releves[0])
```

---
## Partie 1 – Filtrer les données *(20 min)*

**Filtrer** = ne garder que les lignes qui correspondent à un critère.

```python
# Filtre simple : uniquement les données de Juillet
print("=== Relevés de Juillet ===")
for r in releves:
    if r["mois"] == "Juillet":
        print(f"{r['ville']} : {r['temp_moy']}°C, {r['precip_mm']}mm de pluie")
```

```python
# Filtre avec condition numérique : villes avec température > 20°C en Juillet
print("=== Villes avec plus de 20°C en Juillet ===")
for r in releves:
    if r["mois"] == "Juillet" and float(r["temp_moy"]) > 20:
        print(f"{r['ville']} : {r['temp_moy']}°C")
```

```python
# Filtre double : villes très pluvieuses en hiver (Janvier, plus de 80mm)
print("=== Villes très pluvieuses en Janvier (> 80mm) ===")
for r in releves:
    if r["mois"] == "Janvier" and float(r["precip_mm"]) > 80:
        print(f"{r['ville']} ({r['region']}) : {r['precip_mm']}mm")
```

## Partie 2 – Trier les données *(20 min)*

**Trier** = ordonner les lignes selon la valeur d'un descripteur.

```python
# Trier les relevés de Juillet par température (du plus froid au plus chaud)
juillet = [r for r in releves if r["mois"] == "Juillet"]

# La fonction sorted() trie une liste
# key= indique sur quelle valeur trier
juillet_trie = sorted(juillet, key=lambda r: float(r["temp_moy"]))

print("=== Villes de Juillet triées par température (croissant) ===")
for r in juillet_trie:
    print(f"{r['ville']:15} : {r['temp_moy']}°C")
```

```python
# Trier du plus chaud au plus froid : reverse=True
juillet_trie_desc = sorted(juillet, key=lambda r: float(r["temp_moy"]), reverse=True)

print("=== Top 3 des villes les plus chaudes en Juillet ===")
for r in juillet_trie_desc[:3]:  # on prend les 3 premiers
    print(f"  {r['ville']} : {r['temp_moy']}°C")
```

## Partie 3 – Calculs sur les données *(20 min)*

**Calculer** = appliquer des opérations mathématiques sur les valeurs (moyenne, total, min, max...).

```python
# Calcul de la température moyenne de toutes les villes en Juillet
juillet = [r for r in releves if r["mois"] == "Juillet"]

somme_temp = 0
for r in juillet:
    somme_temp = somme_temp + float(r["temp_moy"])

moyenne = somme_temp / len(juillet)
print(f"Température moyenne en Juillet (toutes villes) : {moyenne:.1f}°C")
```

```python
# Trouver la ville la plus ensoleillée et la moins ensoleillée
juillet_trie_soleil = sorted(juillet, key=lambda r: float(r["ensoleillement_h"]))

moins_ensoleille = juillet_trie_soleil[0]
plus_ensoleille = juillet_trie_soleil[-1]

print(f"Moins ensoleillée en Juillet : {moins_ensoleille['ville']} ({moins_ensoleille['ensoleillement_h']}h)")
print(f"Plus ensoleillée en Juillet  : {plus_ensoleille['ville']} ({plus_ensoleille['ensoleillement_h']}h)")
```
